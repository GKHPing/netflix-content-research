# Getting the Iris Dataset

```python
from sklearn import datasets
iris = datasets.load_iris()               
print(iris)                             # raw data of type Bunch
```

```python
print(iris.DESCR)                                  # Description
```

```python
print(iris.data)                                # Features
print(iris.feature_names)                       # Feature Names
```

```python
print(iris.target)                              # Labels
print(iris.target_names)                        # Label names
```

```python
import pandas as pd
df = pd.DataFrame(iris.data, columns=iris.feature_names)  # convert features to 
                                                          # dataframe in Pandas
display(df)
```

```python
target = pd.DataFrame(iris.target)
print(target)
```

```python
df["type"] = target
df
```

```python
## iloc with : returns a series e.g. df.iloc[1]
## iloc without : returns a value  e.g. going into a single row df.iloc[1:2]

print(df.iloc[1,:2])
```

# Getting the Breast Cancer Dataset

```python
breast_cancer = datasets.load_breast_cancer()
print(breast_cancer.data)
print(breast_cancer.feature_names)

print(breast_cancer.target)
print(breast_cancer.target_names)
```

```python
dfCancer = pd.DataFrame(breast_cancer.data, columns=breast_cancer.feature_names)  # convert features to # dataframe in Pandas
display(dfCancer)
```

```python
dfCancer["type"] = pd.DataFrame(breast_cancer.target)
dfCancer["typename"] = pd.DataFrame(breast_cancer.target_names)
dfCancer = dfCancer.drop(columns='typename')
dfCancer
```
